## 4.1 Hexadecimal

Most Western cultures use the decimal system (base-10), but computers use the binary system (base-2). To express data the same way a computer does without dealing with huge chains of `0`s and `1`s, computer scientists use the hexadecimal system (base-16).

Hexadecimal digits range from 0 to 9, followed by `A` through `F` for values 10 through 15. Because a group of four binary digits (bits) has exactly 16 combinations, each group maps perfectly to a single hexadecimal digit.

For example, to convert `01000110101000101011100100111101` to hexadecimal:

1. Group into fours: `0100 0110 1010 0010 1011 1001 0011 1101`
2. Convert each group: `4 6 A 2 B 9 3 D`
3. Prepend `0x`: `0x46A2B93D`

Just like binary and decimal, hexadecimal has place values ($16^0$, $16^1$, $16^2$, etc.).

## 4.2 Pointers

Memory is basically a huge array of 8-bit wide bytes. Every variable is stored in this memory array, and each location has an address.

### 4.2.1 Passing by Reference

Normally, data is passed by value (a copy). Pointers provide an alternative way to pass data between functions by passing the actual variable itself (passing by reference). That means a change made in one function can impact what happens in a different function.

### 4.2.2 Pointer Syntax

Pointers are just addresses. A pointer is a data item whose value is a memory address, and whose type describes the data located at that address.

To extract the address of an existing variable, use the address extraction operator (`&`):

```c
int k = 5;
int* pk = &k;
```

To modify or inspect the location to which a pointer points, use the dereference operator (`*`):

```c
*pk = 35;
```

This assigns `35` to the location `pk` points to, thus modifying `k`.

### 4.2.3 NULL Pointers

The simplest pointer is `NULL`, which points to nothing. If a pointer is not set to a known value immediately, it should always be set to `NULL`.

Dereferencing a `NULL` pointer causes a **segmentation fault**. This is actually good behavior, as it defends against accidental dangerous manipulation of unknown pointers.

## 4.3 Defining Custom Types

The `typedef` keyword creates a shorthand or rewritten name for a data type.

```c
typedef <old name> <new name>;
```

### 4.3.1 Simple Aliases

For example, to alias `unsigned char` to `byte`:

```c
typedef unsigned char byte;
```

To alias `char *` to `string` (which is what `<cs50.h>` does):

```c
typedef char *string;
```

### 4.3.2 Typedef with Structures

`typedef` is especially useful with structures, avoiding the need to type `struct car` every time a variable is declared. A structure can be defined and typedef'd simultaneously:

```c
typedef struct car
{
    // fields...
} car_t;
```

Now, variables can simply be declared as `car_t mycar;`.

## 4.4 Dynamic Memory Allocation

Statically declared memory is set up on the **stack**, which requires knowing exactly how much memory is needed at compile time. Memory created on the fly comes from the **heap**. The stack and heap are two ends of the same giant pool of memory.

### 4.4.1 malloc

The `malloc` function accesses memory on the heap. It takes one argument: the number of bytes requested. It returns a pointer to that contiguous block of memory.

```c
int *px = malloc(sizeof(int));
```

If `malloc` runs out of memory, it returns `NULL`. This condition must always be checked before dereferencing.

### 4.4.2 free

Unlike the stack, dynamically allocated memory is not automatically released back to the system when the function finishes. Failing to release it causes a memory leak.

```c
free(px);
```

Every block of memory allocated with `malloc` must be explicitly freed.

## 4.5 Call Stacks

When a function is called, the system sets aside a chunk of memory called a **stack frame** or **function frame** for it to do its work. More than one stack frame can exist at a time.

### 4.5.1 Push and Pop

Frames are arranged in a stack. The frame for the most-recently called function is always on the top of the stack and is the active frame.

- When a new function is called, a new frame is **pushed** onto the stack.
- When a function finishes, its frame is **popped** off the stack, and the frame immediately below it becomes the new active function, resuming where it left off.

## 4.6 File Pointers

File manipulation functions (found in `<stdio.h>`) allow reading and writing persistent data. This is abstracted by the `FILE` data structure, typically accessed via a file pointer (`FILE *`).

### 4.6.1 fopen and fclose

`fopen` returns a file pointer to a file specified by name, opened in a specific mode (`"r"`, `"w"`, `"a"`).

```c
FILE *ptr = fopen("file.txt", "r");
```

The return value must be checked against `NULL` to ensure the file was successfully opened. When finished, the file must be closed using `fclose(ptr)`.

### 4.6.2 Single Character I/O

`fgetc` reads and returns the next character from the file. `fputc` writes a character to the file.

```c
char ch = fgetc(ptr);
fputc('A', ptr);
```

### 4.6.3 Block I/O

`fread` and `fwrite` read and write blocks of data to a buffer. They expect the buffer pointer, the size of each unit, the quantity of units, and the file pointer.

```c
int arr[10];
fread(arr, sizeof(int), 10, ptr);
fwrite(arr, sizeof(int), 10, ptr);
```
